```python
import pandas as pd
import numpy as np
import sklearn.datasets
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
game_df = pd.read_csv('/content/online_gaming_behavior_dataset - online_gaming_behavior_dataset.csv')
```

```python
game_df.head()
```

```python
game_df['EngagementLevel'].value_counts()
```

```python
game_x = game_df.drop(columns='EngagementLevel',axis=1)
game_y = game_df['EngagementLevel']
```

```python
game_model = LogisticRegression()
```

```python
game_encoder = LabelEncoder()
```

```python
game_df['Gender_Encoded'] = game_encoder.fit_transform(game_df['Gender'])
game_df['Location_Encoded'] = game_encoder.fit_transform(game_df['Location'])
game_df['GameGenre_Encoded'] = game_encoder.fit_transform(game_df['GameGenre'])
game_df['GameDifficulty_Encoded'] = game_encoder.fit_transform(game_df['GameDifficulty'])
```

```python
game_x_train, game_x_test, game_y_train, game_y_test = train_test_split(game_x.drop(['Gender','Location','GameGenre','GameDifficulty'], axis=1),game_y, test_size=0.1, random_state=2)
```

```python
game_model.fit(game_x_train, game_y_train)
```

```python
game_x_pred = game_model.predict(game_x_train)
game_x_acc = accuracy_score(game_x_pred, game_y_train)
```

```python
print(game_x_acc)
```

```python
game_test_pred = game_model.predict(game_x_test)
game_test_acc = accuracy_score(game_test_pred,game_y_test)
```

```python
print(game_test_acc)
```
