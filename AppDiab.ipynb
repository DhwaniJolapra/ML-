```python
X=diab_df.drop(columns='Outcome',axis=1)
Y=diab_df['Outcome']
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

```python
classifier=svm.SVC(kernel='linear')
```

```python
classifier.fit(X_train,Y_train)
```

```python
X_train_pred=classifier.predict(X_train)
```

```python
X_train_acc=accuracy_score(X_train_pred,Y_train)
```

```python
print(X_train_acc)
```

```python
X_test_pred=classifier.predict(X_test)
X_test_acc=accuracy_score(X_test_pred,Y_test)
print(X_test_acc)
```

```python
class model_input(BaseModel):
    Pregnancies:int
    Glucose:int
    BloodPressure:int
    SkinThickness:int
    Insulin:int
    BMI:float
    DiabetesPedigreeFunction:float
    Age:int
```

```python
loaded_model = pickle.load(open('trained_model_dia.pkl', 'rb'))
```

```python
@app.post('/diabetes_prediction')
def dia_pred(input_paramteres: model_input):
    input_data=input_paramteres.json()
    input_dictionary=json.loads(input_data)

    preg=input_dictionary['Pregnancies']
    glu=input_dictionary['Glucose']
    bp=input_dictionary['BloodPressure']
    skin=input_dictionary['SkinThickness']
    insulin=input_dictionary['Insulin']
    bmi=input_dictionary['BMI']
    dp=input_dictionary['DiabetesPedigreeFunction']
    age=input_dictionary['Age']

    input_list=[preg,glu,bp,skin,insulin,bmi,dp,age]
    prediction=loaded_model.predict([input_list])

    if prediction[0] == 0:
        return 'The person is non diabetic'
    else:
        return 'The person is diabetic'
```
