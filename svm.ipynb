```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
```

Data collection and analysis

```python
dhwani=pd.read_csv('/content/diabetes (2) - diabetes (2).csv')
```

```python
dhwani.head()
```

```python
dhwani.shape
```

```python
dhwani.describe()
```

```python
dhwani['Outcome'].value_counts()
```

```python
dhwani.groupby('Outcome').mean()
```

```python
x=dhwani.drop(columns='Outcome',axis=1)
y=dhwani['Outcome']
```

```python
print(x)
```

```python
print(y)
```

Data standardization

```python
scaler=StandardScaler()
```

```python
scaler.fit(x)
x_std=scaler.transform(x)
```

```python
print(x_std)
```

```python
x=x_std
```

Train Test Split

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

Training the model

```python
model=svm.SVC(kernel='linear')
```

Train SVM

```python
model.fit(x_train,y_train)
```

Model Evaluation-accuracy score

```python
x_train_pred=model.predict(x_train)
```

```python
x_train_acc=accuracy_score(x_train_pred,y_train)
```

```python
print(x_train_acc)
```

```python
x_test_pred=model.predict(x_test)
x_test_acc=accuracy_score(x_test_pred,y_test)
print(x_test_acc)
```

```python
class model_input(BaseModel):

  Pregnancies:int
  Glucose:int
  BloodPressure:int
  SkinThickness:int
  Insulin:int
  BMI:float
  DiabetesPedigreeFunction:float
  Age:int
```

```python
loaded_model=pickle.load(open('dia.pkl','rb'))
```

```python
@app.post('/diabetes_prediction')
def dia_pred(input_paramteres: model_input):
    input_data=input_paramteres.json()
    input_dictionary=json.loads(input_data)

    preg=input_dictionary['Pregnancies']
    glu=input_dictionary['Glucose']
    bp=input_dictionary['BloodPressure']
    skin=input_dictionary['SkinThickness']
    insulin=input_dictionary['Insulin']
    bmi=input_dictionary['BMI']
    dp=input_dictionary['DiabetesPedigreeFunction']
    age=input_dictionary['Age']

    input_list=[preg,glu,bp,skin,insulin,bmi,dp,age]
    prediction=loaded_model.predict([input_list])

    if prediction[0] == 0:
        return 'The person is non diabetic'
    else:
        return 'The person is diabetic'
```
