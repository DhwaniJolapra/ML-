```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
generator = pipeline("text2text-generation", model="google/flan-t5-base")
```

```python
def generate_auto_reply(email_text, context="You are a polite and helpful assistant. Reply to this email."):
    prompt = f"{context}\n\nEmail: {email_text}\n\nReply:"
    result = generator(prompt, max_length=150, temperature=0.7)[0]["generated_text"]
    return result
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Input box
email_input = widgets.Textarea(
    value='Would you like to join me for coffee?',
    placeholder='Paste the email content here...',
    description='Email:',
    layout=widgets.Layout(width='100%', height='120px')
)

# Button
reply_button = widgets.Button(
    description='Generate Reply',
    button_style='success'
)

# Output area
output = widgets.Output()

# Callback function
def on_reply_button_click(b):
    with output:
        clear_output()
        email_text = email_input.value.strip()
        if not email_text:
            print("⚠️ Please enter an email message.")
        else:
            reply = generate_auto_reply(email_text)
            print("📬 Auto-Generated Reply:\n")
            print(reply)

reply_button.on_click(on_reply_button_click)

# Display
display(email_input, reply_button, output)
```
