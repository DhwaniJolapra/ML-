```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import nltk
import re
from os import replace
import sklearn
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from xgboost import XGBRegressor
from sklearn import metrics
```

```python
cal=pd.read_csv('/content/calories - calories (1).csv')
```

```python
exersice=pd.read_csv('/content/exercise - exercise (1).csv')
```

```python
exersice.head()
```

```python
exersice.isnull().sum()
```

```python
cal_data=pd.concat([exersice,cal['Calories']],axis=1)
```

```python
cal_data.head()
```

```python
cal_data.replace({"Gender":{'male':0,'female':1}},inplace=True)
```

```python
cal_data.head()
```

```python
x=cal_data.drop(columns=['User_ID','Calories'],axis=1)
y=cal_data['Calories']
```

```python
print(x,y)
```

```python
x_train,x_test,y_train,y_test =train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

```python
model=XGBRegressor()
```

```python
model.fit(x_train,y_train)
```

```python
test_data_pred=model.predict(x_test)
print(test_data_pred)
```

```python
mae=metrics.mean_absolute_error(y_test,test_data_pred)
```
